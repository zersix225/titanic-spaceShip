### Login into kaggle

```python
# Verify Kaggle
from dotenv import load_dotenv
import os
import kaggle, kagglehub

load_dotenv()

kaggle_name = os.getenv("KAGGLE_USERNAME")
kaggle_key = os.getenv("KAGGLE_KEY")

os.environ['KAGGLE_USERNAME'] = kaggle_name or ''
os.environ['KAGGLE_KEY'] = kaggle_key or ''

# Authenticate with Kaggle API
kaggle.api.authenticate()
kagglehub.login()
```

### Store cache var

```python
# Download latest version
cache_path = kagglehub.competition_download('spaceship-titanic')
%store cache_path

print("Path to competition files:", cache_path)
```
