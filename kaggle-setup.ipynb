### Login into kaggle

```python
# Verify Kaggle
from dotenv import load_dotenv
import os
import kaggle, kagglehub

load_dotenv()

kaggle_name = os.getenv("KAGGLE_USERNAME")
kaggle_key = os.getenv("KAGGLE_KEY")

os.environ['KAGGLE_USERNAME'] = kaggle_name or ''
os.environ['KAGGLE_KEY'] = kaggle_key or ''

# Authenticate with Kaggle API
kaggle.api.authenticate()
kagglehub.login()
```

### Store cache var

```python
import os
import shutil

# Download latest version
raw_path = "./data/raw"
clean_path = "./data/clean"

os.makedirs(raw_path, exist_ok=True)
os.makedirs(clean_path, exist_ok=True)

cache_path = kagglehub.competition_download('spaceship-titanic')

for file in os.listdir(cache_path):
    shutil.copy2(
        os.path.join(cache_path, file),
        raw_path
    )
print("Saved to:", os.path.abspath(raw_path))
```
